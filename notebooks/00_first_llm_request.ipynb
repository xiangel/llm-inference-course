## 加载一个小模型

`pipeline` 会一起加载模型权重和与它配套的 Tokenizer。输入是 Python 字符串，输出是生成后的字符串。

```python
from transformers import pipeline

generator = pipeline('text-generation', model='sshleifer/tiny-gpt2')
prompt = 'Python program sends this prompt: '
result = generator(prompt, max_new_tokens=20, do_sample=False)
print(result[0]['generated_text'])
```

## 看看模型实际收到的内容

模型不能直接读取 Python 字符串。下面把 Prompt 交给 Tokenizer，并打印得到的一串整数 ID。

```python
tokenizer = generator.tokenizer
ids = tokenizer.encode(prompt)
print('Prompt:', prompt)
print('token IDs:', ids)
print('token 数:', len(ids))
```
