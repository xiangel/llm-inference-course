## 加载 Tokenizer

这里使用 GPT-2 的公开 Tokenizer。不同模型可能有不同拆分结果，所以它是演示分词机制，不是所有模型的统一规则。

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained('gpt2')

def inspect(text):
    ids = tokenizer.encode(text)
    pieces = [tokenizer.decode([token_id]) for token_id in ids]
    print('原文:', repr(text))
    print('片段:', pieces)
    print('IDs :', ids)
    print('数量:', len(ids), '\n')

inspect('Hello, world!')
inspect('The tokenizer breaks common pieces efficiently.')
inspect('今天学习大模型推理。')
```

## 上下文预算

一次模型调用需要给系统提示词、聊天记录、当前问题和预留回答共同分配 token 预算。下面只做计数，不需要理解 Transformer 内部。

```python
system_prompt = 'You are a helpful assistant.'
history = 'User: Explain KV Cache.\nAssistant: It stores past attention states.'
question = 'User: Give me a short example.'
reserved_for_answer = 200

input_tokens = len(tokenizer.encode(system_prompt + history + question))
print('输入 token:', input_tokens)
print('为回答预留:', reserved_for_answer)
print('总预算:', input_tokens + reserved_for_answer)
```
